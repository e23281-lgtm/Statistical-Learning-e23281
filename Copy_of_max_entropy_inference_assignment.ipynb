# Q. Statistical Mechanics of the Quantum Oscillator

Consider a single quantum harmonic oscillator of angular frequency $\omega$ in thermal equilibrium with a heat bath at temperature $T$. The quantum mechanical energy eigenvalues are given by:

$$E_j = \hbar \omega \left(j + \frac{1}{2}\right), \quad \text{for } j = 0, 1, 2, \dots$$

where $\hbar$ is the reduced Planck constant and $k_B$ is the Boltzmann constant.

**Answer the Following:**

1. **The Partition Function:**
Show that the canonical partition function $Z(T) = \sum_{j=0}^\infty e^{-E_j / (k_B T)}$ forms a convergent geometric series that simplifies to:

$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$


2. **Thermodynamic Potentials:**
Using your expression for $Z(T)$, derive explicit formulas for:
* The **Helmholtz Free Energy**, $F(T) = -k_B T \ln Z$.
* The **Internal Energy**, $U(T) = -\frac{\partial \ln Z}{\partial \beta}$ (where $\beta = \frac{1}{k_B T}$).


3. **Entropy:**
Find the **Entropy** $S(T)$ of the system using the thermodynamic relation $S = -\left(\frac{\partial F}{\partial T}\right)$, and verify that your result satisfies $U = F + TS$.
4. **Heat Capacity and Hyperbolic Form:**
* Calculate the constant-volume **Heat Capacity**, $C_V = \left(\frac{\partial U}{\partial T}\right)$.
* Use the hyperbolic identity $\sinh(x) = \frac{e^x - e^{-x}}{2}$ to prove that your expression for $C_V$ can be compactly rewritten as:

$$C_V = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$




5. **Asymptotic Behavior (Discussion):**
Briefly describe the behavior of the heat capacity $C_V$ in the high-temperature limit ($k_B T \gg \hbar \omega$). Does it match the classical equipartition theorem prediction?

###1.
 begin by substituting the explicit expression for the quantum mechanical energy eigenvalues ($E_j$) directly into the definition of the canonical partition function:

$$Z(T) = \sum_{j=0}^{\infty} e^{-\frac{E_j}{k_B T}} = \sum_{j=0}^{\infty} e^{-\frac{\hbar\omega\left(j + \frac{1}{2}\right)}{k_B T}}$$

Using the algebraic properties of exponents ($e^{a+b} = e^a \cdot e^b$), we separate the index-independent zero-point energy term from the summation:

$$Z(T) = \sum_{j=0}^{\infty} e^{-\frac{\hbar\omega \cdot j}{k_B T}} \cdot e^{-\frac{\hbar\omega}{2k_B T}}$$

Because the baseline zero-point term $e^{-\frac{\hbar\omega}{2k_B T}}$ does not depend on the summation index $j$, it can be factored completely outside the sum:

$$Z(T) = e^{-\frac{\hbar\omega}{2k_B T}} \sum_{j=0}^{\infty} e^{-\frac{\hbar\omega \cdot j}{k_B T}}$$

Now, we rewrite the term inside the summation as a power of $j$:

$$Z(T) = e^{-\frac{\hbar\omega}{2k_B T}} \sum_{j=0}^{\infty} \left(e^{-\frac{\hbar\omega}{k_B T}}\right)^j$$

#### **Verification of Geometric Series Convergence:**
This expression perfectly matches the standard structural definition of an infinite geometric series:
$$\sum_{j=0}^{\infty} r^j = 1 + r + r^2 + r^3 + \dots$$
where the common ratio is $r = e^{-\frac{\hbar\omega}{k_B T}}$.

Because $\hbar, \omega, k_B,$ and $T$ are strictly positive real physical parameters, the exponent is negative, which guarantees that:
$$0 < e^{-\frac{\hbar\omega}{k_B T}} < 1 \implies |r| < 1$$

Since the absolute value of the common ratio is strictly less than 1, the infinite geometric series is rigorously proven to converge. Its mathematical sum resolves using the standard formula $\frac{1}{1-r}$:

$$\sum_{j=0}^{\infty} \left(e^{-\frac{\hbar\omega}{k_B T}}\right)^j = \frac{1}{1 - e^{-\frac{\hbar\omega}{k_B T}}}$$

#### **Final Simplification:**
Multiplying the factored constant back by the sum of the series yields the final desired result:

$$Z(T) = e^{-\frac{\hbar\omega}{2k_B T}} \cdot \left( \frac{1}{1 - e^{-\frac{\hbar\omega}{k_B T}}} \right) = \frac{e^{-\frac{\hbar\omega}{2k_B T}}}{1 - e^{-\frac{\hbar\omega}{k_B T}}}$$

This completes the mathematical proof.

###2.

#### **The Helmholtz Free Energy, $F(T)$**
$$F(T) = -k_B T \ln Z$$
$$F(T) = -k_B T \ln \left( \frac{e^{-\frac{\hbar\omega}{2k_B T}}}{1 - e^{-\frac{\hbar\omega}{k_B T}}} \right) = -k_B T \left[ -\frac{\hbar\omega}{2k_B T} - \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) \right]$$
$$F(T) = \frac{\hbar\omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) \quad \text{[cite: 1]}$$



#### **The Internal Energy, $U(T)$**
Express $\ln Z$ using $\beta = \frac{1}{k_B T}$[cite: 1]:
$$\ln Z = -\frac{\beta\hbar\omega}{2} - \ln\left(1 - e^{-\beta\hbar\omega}\right)$$

Evaluate $U = -\frac{\partial \ln Z}{\partial \beta}$[cite: 1]:
$$U = -\left[ -\frac{\hbar\omega}{2} - \frac{1}{1 - e^{-\beta\hbar\omega}} \cdot \left(-e^{-\beta\hbar\omega}\right) \cdot (-\hbar\omega) \right] = \frac{\hbar\omega}{2} + \frac{\hbar\omega e^{-\beta\hbar\omega}}{1 - e^{-\beta\hbar\omega}}$$

Divide the second term by $e^{-\beta\hbar\omega}$ and restore $k_B T$[cite: 1]:
$$U(T) = \hbar\omega \left( \frac{1}{2} + \frac{1}{e^{\frac{\hbar\omega}{k_B T}} - 1} \right) \quad \text{[cite: 1]}$$

### 3. Entropy

Differentiating our derived Helmholtz Free Energy function with respect to $T$[cite: 1]:
$$S(T) = -\frac{\partial F}{\partial T} = -\frac{\partial}{\partial T}\left[ \frac{\hbar\omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) \right] \quad \text{[cite: 1]}$$
$$S(T) = -k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) - k_B T \cdot \frac{1}{1 - e^{-\frac{\hbar\omega}{k_B T}}} \cdot \left(-e^{-\frac{\hbar\omega}{k_B T}}\right) \cdot \left(-\frac{\hbar\omega}{k_B T^2}\right) \quad \text{[cite: 1]}$$
$$S(T) = -k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) + \frac{\hbar\omega}{T}\frac{1}{e^{\frac{\hbar\omega}{k_B T}} - 1} \quad \text{[cite: 1]}$$

#### **Verification ($U = F + TS$):**
$$F + TS = \left[ \frac{\hbar\omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) \right] + T \left[ -k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) + \frac{\hbar\omega}{T}\frac{1}{e^{\frac{\hbar\omega}{k_B T}} - 1} \right] \quad \text{[cite: 1]}$$
$$F + TS = \frac{\hbar\omega}{2} + \frac{\hbar\omega}{e^{\frac{\hbar\omega}{k_B T}} - 1} = \hbar\omega\left(\frac{1}{2} + \frac{1}{e^{\frac{\hbar\omega}{k_B T}} - 1}\right) = U(T) \quad \text{[cite: 1]}$$
The identity is verified[cite: 1].



### 4. Heat Capacity and Hyperbolic Form

Differentiating internal energy with respect to $T$[cite: 1]:
$$C_V = \frac{\partial U}{\partial T} = \hbar\omega \cdot \left[ -1\left(e^{\frac{\hbar\omega}{k_B T}} - 1\right)^{-2} \cdot e^{\frac{\hbar\omega}{k_B T}} \cdot \left(-\frac{\hbar\omega}{k_B T^2}\right) \right] \quad \text{[cite: 1]}$$
$$C_V = k_B \left(\frac{\hbar\omega}{k_B T}\right)^2 \frac{e^{\frac{\hbar\omega}{k_B T}}}{\left(e^{\frac{\hbar\omega}{k_B T}} - 1\right)^2} \quad \text{[cite: 1]}$$

To convert to hyperbolic form, divide both the numerator and denominator by $e^{\frac{\hbar\omega}{k_B T}}$[cite: 1]:
$$C_V = k_B \left(\frac{\hbar\omega}{k_B T}\right)^2 \frac{1}{\left(e^{\frac{\hbar\omega}{2k_B T}} - e^{-\frac{\hbar\omega}{2k_B T}}\right)^2} \quad \text{[cite: 1]}$$

Since $e^{\frac{\hbar\omega}{2k_B T}} - e^{-\frac{\hbar\omega}{2k_B T}} = 2\sinh\left(\frac{\hbar\omega}{2k_B T}\right)$, squaring it gives $4\sinh^2\left(\frac{\hbar\omega}{2k_B T}\right)$[cite: 1]:
$$C_V = k_B \left(\frac{\hbar\omega}{k_B T}\right)^2 \frac{1}{4\sinh^2\left(\frac{\hbar\omega}{2k_B T}\right)} = k_B \left(\frac{\hbar\omega}{2k_B T}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar\omega}{2k_B T}\right)} \quad \text{[cite: 1]}$$



### 5. Asymptotic Behavior

In the high-temperature limit ($k_B T \gg \hbar\omega$), the term $x = \frac{\hbar\omega}{2k_B T} \rightarrow 0$. Using the small-angle Taylor approximation $\sinh(x) \approx x$:
$$\sinh^2\left(\frac{\hbar\omega}{2k_B T}\right) \approx \left(\frac{\hbar\omega}{2k_B T}\right)^2$$

Substituting this back into the $C_V$ equation yields:
$$C_V \approx k_B \left(\frac{\hbar\omega}{2k_B T}\right)^2 \cdot \frac{1}{\left(\frac{\hbar\omega}{2k_B T}\right)^2} = k_B$$

**Conclusion:** Yes, this matches the classical equipartition theorem prediction. A 1D harmonic oscillator has two quadratic degrees of freedom (one kinetic, one potential), each contributing $\frac{1}{2}k_B$, yielding a total high-temperature limit of $C_V = k_B$.



The mathematically rigorous continuous Shannon-Jaynes entropy functional is defined as:
$$\hat{H}_c[p] = -\int_{-\infty}^{\infty} p(x) \ln\left(\frac{p(x)}{m(x)}\right) dx$$

Since the baseline measure is uniform ($m(x) = 1$), this simplifies directly to the standard differential entropy equation. To find the unique global maximum distribution $p^*(x)$ under the constraints, we construct the functional Lagrangian using the calculus of variations:
$$\mathcal{L}[p] = -\int_{-\infty}^{\infty} p(x) \ln p(x) \, dx - \gamma \left( \int_{-\infty}^{\infty} p(x) \, dx - 1 \right) - \lambda_1 \left( \int_{-\infty}^{\infty} p(x) \cdot x \, dx - \mu \right) - \lambda_2 \left( \int_{-\infty}^{\infty} p(x) \cdot (x - \mu)^2 \, dx - \sigma^2 \right)$$

Taking the functional derivative (Euler-Lagrange derivative) with respect to $p(x)$ and setting it to zero to locate the stationary point:
$$\frac{\delta \mathcal{L}}{\delta p(x)} = -\ln p(x) - 1 - \gamma - \lambda_1 x - \lambda_2 (x - \mu)^2 = 0$$

Isolating the target density function $p^*(x)$ yields the exponential canonical distribution form:
$$p^*(x) = \frac{1}{Z} \exp\left\{ -\lambda_1 x - \lambda_2(x - \mu)^2 \right\}$$
where $Z = \exp(1 + \gamma)$ serves as the continuous normalization partition function constant.

#### **Evaluating the Multipliers and Integrals:**
1. **The First Multiplier ($\lambda_1$):** Because the distribution must be perfectly symmetric around its given mean $\mu$ to ensure $\mathbb{E}[X] = \mu$ without shifting the coordinate frame, the linear multiplier vanishes cleanly:
   $$\lambda_1 = 0$$

2. **The Normalization Constant ($Z$):** Substituting $\lambda_1 = 0$, the constraint $\int_{-\infty}^{\infty} p^*(x) \, dx = 1$ requires:
   $$Z = \int_{-\infty}^{\infty} \exp\left\{ -\lambda_2(x - \mu)^2 \right\} dx$$
   Using the standard Gaussian integral solution $\int_{-\infty}^{\infty} e^{-a(x-b)^2} dx = \sqrt{\frac{\pi}{a}}$, we get:
   $$Z = \sqrt{\frac{\pi}{\lambda_2}}$$

3. **The Second Multiplier ($\lambda_2$):** Plugging $p^*(x) = \frac{1}{Z} \exp\left\{ -\lambda_2(x - \mu)^2 \right\}$ directly into our variance constraint equation:
   $$\int_{-\infty}^{\infty} \frac{1}{Z} \exp\left\{ -\lambda_2(x - \mu)^2 \right\} (x - \mu)^2 \, dx = \sigma^2$$
   Evaluating this integral via integration by parts yields $\frac{1}{2\lambda_2} = \sigma^2$. Isolating our parameters:
   $$\lambda_2 = \frac{1}{2\sigma^2}, \quad Z = \sqrt{2\pi\sigma^2}$$

#### **Final Density Function:**
Substituting the solved Lagrange multipliers back into our distribution functional directly provides:
$$p^*(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

**Conclusion:** This completes the derivation. If you model an unknown continuous system knowing nothing but its mean and variance, assuming any shape other than a Gaussian distribution actively introduces arbitrary, unproven human bias into your calculation.

# Q. Asset Pricing and Portfolio Optimization.



In standard finance, models often assume asset returns follow a perfect Normal (Gaussian) distribution. However, real-world markets experience extreme events, leading to "fat tails" and skewed returns. MaxEnt allows an analyst to infer the **least-biased probability distribution** of an asset's future price based *only* on what the market currently tells us, without forcing it into a rigid Gaussian shape.

**Background:**

You are a quantitative portfolio manager evaluating a highly volatile tech stock, **Asset X**, currently trading at $\$100$. You want to determine the risk-neutral probability distribution of its price, $S_T$, in exactly one month. This distribution will allow you to correctly price exotic derivatives and optimize your portfolio's risk exposure.

**Available Market Data (The Constraints):**

You do not know the true distribution, but you can observe the following macroeconomic aggregates and market-clearing prices from the options exchange:

1. **The Risk-Free Growth:** Based on the risk-free interest rate, the expected future price of the asset must match the forward price:

$$\mathbb{E}[S_T] = \$102$$


2. **The Market Volatility:** Based on standard historical variance, the expected squared price is constrained by:

$$\mathbb{E}[S_T^2] = 10,600$$


3. **Asymmetric Market Sentiment (The "Skew"):** Due to heavy trading on out-of-the-money put options, the market is pricing in a strong downside risk. The expected value of a specific logarithmic penalty (which captures downward crash risk) is constrained to:

$$\mathbb{E}[\ln(S_T)] = 4.60$$



**Objective:**

Formulate and solve the Maximum Entropy problem to find the continuous probability density function $p(S_T)$ for $S_T > 0$.
