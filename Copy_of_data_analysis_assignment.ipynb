QUESTION 1: Bias in MLE Covariance Matrix
==========================================

The MLE estimator is:
  Σ_MLE = (1/n) Σ (Xi - μ̂)(Xi - μ̂)ᵀ

Expanding deviations around true mean μ:
  Xi - μ̂ = (Xi - μ) - (μ̂ - μ)

After substitution and expanding the product:
  Σ(Xi - μ̂)(Xi - μ̂)ᵀ = Σ(Xi - μ)(Xi - μ)ᵀ - n(μ̂ - μ)(μ̂ - μ)ᵀ

Taking expectation:
  E[Σ_MLE] = (1/n)[nΣ - n·(1/n)Σ] = (n-1)/n · Σ

Since E[Σ_MLE] = (n-1)/n · Σ ≠ Σ  →  the MLE is BIASED (downward).

Bessel's Correction:
  S = n/(n-1) · Σ_MLE = 1/(n-1) · Σ(Xi - μ̂)(Xi - μ̂)ᵀ

  E[S] = n/(n-1) · (n-1)/n · Σ = Σ  ✓  (Unbiased)

The bias occurs because deviations are computed against the sample mean
(which is closer to the data than the true mean), artificially shrinking
the estimated spread.

Slutsky's Theorem (Rigorous Statement):
========================================
Let {Xn} and {Yn} be sequences of random variables/vectors.
If:
  Xn →(d) X   (converges in distribution)
  Yn →(p) c   (converges in probability to a constant c)

Then:
  Xn + Yn →(d) X + c
  Xn · Yn →(d) c · X
  Xn / Yn →(d) X / c   (provided c ≠ 0)


Justification for Substituting S for Σ:
=========================================

Step 1 — Lindeberg-Lévy Multivariate CLT:
  √n (μ̂ - μ) →(d) N(0, Σ)

  Equivalently:  Σ^(-1/2) · √n(μ̂ - μ) →(d) N(0, I)

Step 2 — Weak Law of Large Numbers (WLLN):
  S is a continuous, consistent estimator of Σ.
  Therefore as n → ∞:
    S →(p) Σ  ⟹  S^(-1/2) →(p) Σ^(-1/2)

Step 3 — Apply Slutsky's Theorem:
  Multiply the distribution-converging sequence by the
  probability-converging sequence:

    S^(-1/2) · √n(μ̂ - μ) →(d) Σ^(-1/2) · N(0,Σ) = N(0, I)

Step 4 — Re-scale back to empirical metric:
  √n(μ̂ - μ) ~ N(0, S)
  ⟹  μ̂ ~ N(μ, (1/n)S)

This mathematically validates substituting S for Σ without corrupting
the asymptotic diagnostic envelope.

````python
import numpy as np
import pandas as pd
import scipy.stats as stats

# --- DO NOT MODIFY: Synthetic Strain Sensor Data Generator ---
np.random.seed(42)
n_samples = 5000
n_features = 3

base_data = np.random.multivariate_normal(
    mean=[0.5, -0.2, 1.1],
    cov=[[0.09, 0.02, 0.01],
         [0.02, 0.06, 0.03],
         [0.01, 0.03, 0.05]],
    size=n_samples
)

base_data[4000:, 0] += 0.015  # Sensor 1 drift
base_data[4000:, 2] -= 0.010  # Sensor 3 drift

df_strain = pd.DataFrame(base_data, columns=['Strain_Ch1', 'Strain_Ch2', 'Strain_Ch3'])
# ----------------------------------------------------------------

def verify_first_moment_homogeneity(df: pd.DataFrame, g_chunks: int = 5) -> dict:
    """
    Partitions the dataset into g chunks and evaluates first-moment homogeneity
    via Wilks' Lambda and Bartlett's Chi-Square asymptotic transformation.
    """
    n, m = df.shape
    chunk_size = n // g_chunks

    # 1. Split df into g equal consecutive blocks
    chunks = [df.iloc[j * chunk_size:(j + 1) * chunk_size].values
              for j in range(g_chunks)]

    # 2. Compute global mean and individual chunk means
    global_mean = df.values.mean(axis=0)
    chunk_means = [c.mean(axis=0) for c in chunks]
    chunk_sizes = [len(c) for c in chunks]

    # 3. Calculate Within-Chunk Matrix (W) and Between-Chunk Matrix (B)
    W = np.zeros((m, m))
    for c in chunks:
        diff = c - c.mean(axis=0)
        W += diff.T @ diff

    B = np.zeros((m, m))
    for nj, mu_j in zip(chunk_sizes, chunk_means):
        diff = (mu_j - global_mean).reshape(-1, 1)
        B += nj * (diff @ diff.T)

    # 4. Compute Wilks' Lambda
    det_W = np.linalg.det(W)
    det_BW = np.linalg.det(B + W)
    wilks_lambda = det_W / det_BW

    # 5. Bartlett's Chi-Square approximation
    g = g_chunks
    chi2_calc = -(n - 1 - (m + g) / 2) * np.log(wilks_lambda)

    # Degrees of freedom: m*(g-1)
    df_chi2 = m * (g - 1)

    # p-value
    p_value = 1 - stats.chi2.cdf(chi2_calc, df=df_chi2)

    print("=" * 55)
    print("   MANOVA First-Moment Homogeneity Test Results")
    print("=" * 55)
    print(f"  Wilks' Lambda (Λ)         : {wilks_lambda:.6f}")
    print(f"  Bartlett Chi² Statistic   : {chi2_calc:.4f}")
    print(f"  Degrees of Freedom        : {df_chi2}")
    print(f"  p-value                   : {p_value:.6e}")
    print("-" * 55)
    if p_value < 0.05:
        print("  CONCLUSION: REJECT H₀ at α=0.05")
        print("  ➜ The baseline CENTER HAS SHIFTED over time.")
        print("  ➜ First moment is NOT homogeneous.")
    else:
        print("  CONCLUSION: FAIL TO REJECT H₀ at α=0.05")
        print("  ➜ First moment IS homogeneous.")
    print("=" * 55)

    return {
        "wilks_lambda": wilks_lambda,
        "chi2_statistic": chi2_calc,
        "degrees_of_freedom": df_chi2,
        "p_value": p_value
    }

# Execute
results = verify_first_moment_homogeneity(df_strain, g_chunks=5)
print(results)
````

Complete the Python class method below inside Google Colab. The function must:

3. Loop through every possible subspace cutoff value $k \in \{1, 2, 3\}$, calculating the sample vector array for Hotelling’s $T^2$ ($T^2_i = \sum_{j=1}^k \frac{z_{i,j}^2}{\widehat{\lambda}_j}$) and the $Q$ statistic ($Q_i = \sum_{j=k+1}^m z_{i,j}^2$).

````python
import numpy as np
import pandas as pd

# ============================================================
# PCA PART B: NUMERICAL VERIFICATION
# ============================================================

np.random.seed(42)
n, m = 3000, 4

# Generate synthetic baseline data (4 sensors, 3000 snapshots)
true_cov = np.array([
    [1.00, 0.80, 0.30, 0.10],
    [0.80, 1.00, 0.25, 0.08],
    [0.30, 0.25, 1.00, 0.15],
    [0.10, 0.08, 0.15, 1.00]
])
X_baseline = np.random.multivariate_normal(mean=np.zeros(m),
                                            cov=true_cov,
                                            size=n)
df_baseline = pd.DataFrame(X_baseline,
                            columns=['Sensor_1','Sensor_2','Sensor_3','Sensor_4'])

def compute_pca_diagnostics(df: pd.DataFrame):
    """
    PCA pipeline computing T² and Q statistics for all k in {1,2,3}.
    """
    X = df.values.copy()
    n_obs, m_feat = X.shape

    # 1. Centre the data
    mu_hat = np.mean(X, axis=0)
    X_centered = X - mu_hat

    # 2. Unbiased sample covariance matrix
    S = np.cov(X, rowvar=False, ddof=1)

    # 3. Spectral decomposition (eigh for symmetric matrices)
    eigenvalues, eigenvectors = np.linalg.eigh(S)
    idx = np.argsort(eigenvalues)[::-1]          # descending order
    lambda_hat = eigenvalues[idx]
    P_hat = eigenvectors[:, idx]
    lambda_hat = np.clip(lambda_hat, 1e-15, None)

    # 4. Project all samples onto PC space
    Z = X_centered @ P_hat                       # shape (n, m)

    # 5. Loop through k in {1, 2, 3}
    k_range = list(range(1, m_feat))
    mean_T2 = []
    mean_Q  = []

    print("=" * 55)
    print("   PCA Diagnostics: T² and Q per Truncation k")
    print("=" * 55)

    for k in k_range:
        # Hotelling's T²
        Z_k      = Z[:, :k]
        lam_k    = lambda_hat[:k]
        T2_i     = np.sum((Z_k ** 2) / lam_k, axis=1)

        # Q (SPE) statistic
        Z_resid  = Z[:, k:]
        Q_i      = np.sum(Z_resid ** 2, axis=1)

        mean_T2.append(np.mean(T2_i))
        mean_Q.append(np.mean(Q_i))

        print(f"  k={k}  |  Mean T²  = {np.mean(T2_i):.4f}"
              f"  |  Mean Q = {np.mean(Q_i):.4f}")

    print("=" * 55)
    print("\nEigenvalues (λ̂₁ ≥ ... ≥ λ̂ₘ):", np.round(lambda_hat, 4))
    total_var = np.sum(lambda_hat)
    cum_var   = np.cumsum(lambda_hat) / total_var * 100
    print("Cumulative Explained Variance (%):", np.round(cum_var, 2))

    return {
        "eigenvalues"    : lambda_hat,
        "eigenvectors_P" : P_hat,
        "mean_T2_vs_k"   : mean_T2,
        "mean_Q_vs_k"    : mean_Q,
        "k_range"        : k_range
    }

pca_results = compute_pca_diagnostics(df_baseline)
````

#1.
 The Fundamental Equation of Factor AnalysisThe generative model for the $i$-th realization vector is defined as $Z_{i} = \Lambda F_{i} + \epsilon_{i}$. We evaluate the total population correlation matrix $R$ by expanding this generative formula:
  $$R = E[Z_{i}Z_{i}^{T}] = E[(\Lambda F_{i} + \epsilon_{i})(\Lambda F_{i} + \epsilon_{i})^{T}]$$$$R = \Lambda E[F_{i}F_{i}^{T}]\Lambda^{T} + \Lambda E[F_{i}\epsilon_{i}^{T}] + E[\epsilon_{i}F_{i}^{T}]\Lambda^{T} + E[\epsilon_{i}\epsilon_{i}^{T}]$$  We apply the system constraints: the common factors are orthogonal ($E[F_{i}F_{i}^{T}] = I_{k \times k}$), the unique errors are diagonal ($E[\epsilon_{i}\epsilon_{i}^{T}] = \Psi$), and the subspaces are orthogonal ($E[\epsilon_{i}F_{i}^{T}] = 0_{m \times k}$). Substituting these into the expansion cleans it perfectly to yield the Fundamental Equation of Factor Analysis:
  $$R = \Lambda\Lambda^{T} + \Psi$$  Focusing on the $j$-th diagonal entry, the total unit variance splits into two spaces: $E[Z_{i,j}^{2}] = \sum_{r=1}^{k}\lambda_{j,r}^{2} + \varphi_{j}^{2} = 1.0$.  Communality ($h_{j}^{2}$): Defined as $h_{j}^{2} \triangleq \sum_{r=1}^{k}\lambda_{j,r}^{2}$. This represents the portion of a sensor's variance explained collectively by the $k$ shared structural factors.  Uniqueness ($\varphi_{j}^{2}$): Defined as $\varphi_{j}^{2} \triangleq 1.0 - h_{j}^{2} = E[\epsilon_{i,j}^{2}]$. This is the remaining portion of variance completely isolated to that specific sensor, representing a combination of localized non-structural anomalies and electronic white noise.
  
#2.
   Orthogonal Rotation and Geometric SubspacesConceptual Difference: While Principal Component Analysis (PCA) merely rotates the empirical coordinate system to maximize total variance, Factor Analysis models the internal correlation structure by partitioning the variance into shared and independent subspaces. A raw PCA loading vector forces maximum global variance, whereas a Varimax-rotated factor loading vector enforces a "simple structure" that clearly maps individual sensors to specific, isolated physical modes.  Matrix Stability: If we apply an orthogonal rotation matrix $T$ (where $TT^T = I$) to the loading matrix such that $\Lambda^* = \Lambda T$, we can substitute it into the global covariance approximation:$$R \approx (\Lambda T)(\Lambda T)^{T} + \Psi = \Lambda(TT^{T})\Lambda^{T} + \Psi = \Lambda(I)\Lambda^{T} + \Psi = \Lambda\Lambda^{T} + \Psi$$Because the entire reconstructed covariance matrix $R$ is mathematically preserved, the individual communalities ($h_{j}^{2}$)—which are the row-wise sum of squares of $\Lambda$—remain completely unchanged by the rotation.3. Thomson's Regression Method (Conditional Expectations)To find the conditional expectation, we first examine how the observable variables and hidden factors vary together by constructing the partitioned joint vector $Y_{i} = [Z_{i}^{T}, F_{i}^{T}]^{T}$. The cross-covariance block is:$$E[Z_{i}F_{i}^{T}] = E[(\Lambda F_{i} + \epsilon_{i})F_{i}^{T}] = \Lambda E[F_{i}F_{i}^{T}] + E[\epsilon_{i}F_{i}^{T}] = \Lambda I + 0 = \Lambda$$  Knowing $E[Z_{i}Z_{i}^{T}] = R$ and $E[F_{i}F_{i}^{T}] = I_{k \times k}$, the joint population covariance matrix perfectly expands to:
  $$\Sigma_{YY} = \begin{bmatrix} R & \Lambda \\ \Lambda^{T} & I_{k \times k} \end{bmatrix}$$Under the assumption of a Multivariate Normal distribution, the conditional expectation of the second slot (the factors) given a realization of the first slot (the observed data) utilizes the classical linear regression projection formula: $E[X_{2} | x_{1}] = \mu_{2} + \Sigma_{21}\Sigma_{11}^{-1}(x_{1} - \mu_{1})$.
Mapping our specific partitions into this theorem gives $\Sigma_{21} = \Lambda^{T}$ and $\Sigma_{11} = R$. Substituting these values directly yields Thomson's final score matrix equation:
  $$f_{i} = 0 + \Lambda^{T}R^{-1}(z_{i} - 0) \implies f_{i} = \Lambda^{T}R^{-1}z_{i}$$

````python
import numpy as np
import pandas as pd
from sklearn.decomposition import FactorAnalysis
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ============================================================
# DATA GENERATOR
# ============================================================
np.random.seed(42)
n_samples = 2500

f1 = np.random.normal(0, 1, n_samples)
f2 = np.random.normal(0, 1, n_samples)

s1 = 0.85 * f1 + 0.10 * f2 + np.random.normal(0, 0.3,  n_samples)
s2 = 0.80 * f1 + 0.15 * f2 + np.random.normal(0, 0.35, n_samples)
s3 = 0.12 * f1 + 0.90 * f2 + np.random.normal(0, 0.25, n_samples)
s4 = 0.02 * f1 + 0.05 * f2 + np.random.normal(0, 1.40, n_samples)

df_asset = pd.DataFrame(
    data=np.vstack([s1, s2, s3, s4]).T,
    columns=['Sensor_1', 'Sensor_2', 'Sensor_3', 'Sensor_4']
)

# ============================================================
# STEP 1 — Z-SCORE STANDARDISATION WITH GUARDRAILS
# ============================================================
def standardise(df):
    X = df.select_dtypes(include=[np.number]).values.copy()
    n, m = X.shape

    # Dimensionality guardrails
    if n <= m:
        raise ValueError("Snapshot count n must exceed channel count m.")

    mu  = np.mean(X, axis=0)
    std = np.std(X,  axis=0, ddof=1)
    std[std == 0] = 1e-15          # zero-variance guard

    Z = (X - mu) / std
    return Z, mu, std

Z, mu_hat, std_hat = standardise(df_asset)
print(f"Z-score matrix shape : {Z.shape}")
print(f"Mean of Z (should ≈ 0): {np.mean(Z, axis=0).round(4)}")
print(f"Std  of Z (should ≈ 1): {np.std(Z,  axis=0, ddof=1).round(4)}")

# ============================================================
# STEP 2 — FIT FACTOR ANALYSIS WITH VARIMAX ROTATION (k=2)
# ============================================================
k = 2   # number of latent factors

# Dimensionality constraint check
m = Z.shape[1]
if k >= m:
    raise ValueError(f"k={k} must be strictly less than m={m}.")

fa = FactorAnalysis(n_components=k, rotation='varimax', random_state=42)
fa.fit(Z)

lambda_matrix = fa.components_.T    # shape (m, k) — rotated loadings
uniqueness    = fa.noise_variance_  # φ² vector

# ============================================================
# STEP 3 — COMMUNALITY AND UNIQUENESS
# ============================================================
communality = np.sum(lambda_matrix ** 2, axis=1)   # h²

print("\n--- Variance Decomposition ---")
sensor_labels = list(df_asset.columns)
for i, col in enumerate(sensor_labels):
    print(f"  {col}: h²={communality[i]:.4f}  φ²={uniqueness[i]:.4f}")

# ============================================================
# STEP 4 — THOMSON'S FACTOR SCORES: F = Z · R⁻¹ · Λ_rotated
# ============================================================
F_scores = fa.transform(Z)          # shape (n, k)
print(f"\nFactor Scores shape : {F_scores.shape}")

# ============================================================
# STEP 5 — DIAGNOSTIC TELEMETRY
# ============================================================
factor_variances = np.var(F_scores, axis=0, ddof=1)

print("\n" + "=" * 55)
print("   Factor Analysis Diagnostic Telemetry")
print("=" * 55)
print(f"  Average System Communality % = {np.mean(communality)*100:.2f}%")
print(f"  Average System Uniqueness  % = {np.mean(uniqueness)*100:.2f}%")
for j in range(k):
    print(f"  Factor {j+1} Empirical Variance : {factor_variances[j]:.4f}")
print("=" * 55)

# ============================================================
# STEP 6 — 2x2 DIAGNOSTIC DASHBOARD
# ============================================================
factor_labels = [f"Factor {j+1}" for j in range(k)]

fig = make_subplots(
    rows=2, cols=2,
    horizontal_spacing=0.24,
    vertical_spacing=0.28,
    subplot_titles=(
        "Structural Loadings Matrix |λ_(j,r)|",
        "Variance Partitioning (Communality vs Uniqueness)",
        "Sensor Uniqueness Noise Floor (φ²)",
        "Latent Factor Scores Empirical Variance"
    )
)

# --- Subplot (1,1): Loadings Heatmap ---
fig.add_trace(go.Heatmap(
    z=np.abs(lambda_matrix),
    x=factor_labels,
    y=sensor_labels,
    colorscale='YlOrRd',
    colorbar=dict(
        title="Sensitivity",
        x=-0.15, len=0.38,
        y=0.78, yanchor="middle", xanchor="right"
    ),
    showscale=True, showlegend=False
), row=1, col=1)
fig.update_xaxes(title_text="Latent Structures", row=1, col=1)

# --- Subplot (1,2): Stacked Horizontal Bar ---
fig.add_trace(go.Bar(
    y=sensor_labels, x=communality * 100,
    name="Communality h² (%)",
    orientation='h',
    marker_color='#1f77b4'
), row=1, col=2)
fig.add_trace(go.Bar(
    y=sensor_labels, x=uniqueness * 100,
    name="Uniqueness φ² (%)",
    orientation='h',
    marker_color='#ff7f0e'
), row=1, col=2)
fig.update_layout(barmode='stack')
fig.update_xaxes(title_text="Variance Allocation (%)",
                  range=[0, 100], row=1, col=2)

# --- Subplot (2,1): Uniqueness Line ---
fig.add_trace(go.Scatter(
    x=sensor_labels, y=uniqueness,
    mode='lines+markers',
    name="Uniqueness φ²",
    line=dict(color='#d62728', width=2, dash='dashdot'),
    marker=dict(symbol='x', size=10)
), row=2, col=1)
fig.update_yaxes(range=[-0.05, 1.05], row=2, col=1)
fig.update_xaxes(title_text="Monitored Channels",
                  tickangle=25, row=2, col=1)

# --- Subplot (2,2): Factor Variances Bar ---
fig.add_trace(go.Bar(
    x=factor_labels, y=factor_variances,
    name="Factor Variance",
    marker=dict(color='#2ca02c',
                 line=dict(color='black', width=0.5))
), row=2, col=2)
fig.update_yaxes(title_text="Variance Level", row=2, col=2)
fig.update_xaxes(title_text="Latent Vectors",  row=2, col=2)

# --- Global Layout ---
fig.update_layout(
    title="Factor Analysis (FA) Latent Subspace Diagnostics",
    template="plotly_white",
    showlegend=True,
    legend=dict(
        orientation="h", yanchor="bottom",
        y=1.02, xanchor="center", x=0.5,
        itemwidth=40, itemsizing="constant"
    ),
    margin=dict(t=150, b=60, l=140, r=80),
    height=750, width=1250
)
fig.show()
````
