#Q1) Prediction of fraudulent transactions

A bank monitors credit-card transactions for possible fraud.

For each transaction, let the sample space be

$$
\Omega=\{\omega_1,\omega_2,\omega_3,\omega_4\},
$$

where:

* $\omega_1$: low-risk customer, legitimate transaction,
* $\omega_2$: low-risk customer, fraudulent transaction,
* $\omega_3$: high-risk customer, legitimate transaction,
* $\omega_4$: high-risk customer, fraudulent transaction.

Assume the probabilities are

$$
P(\{\omega_1\})=0.50,\qquad
P(\{\omega_2\})=0.10,\qquad
P(\{\omega_3\})=0.20,\qquad
P(\{\omega_4\})=0.20.
$$

Let $X:\Omega\to\mathbb R$ denote the fraud indicator:

$$
X(\omega_1)=0,\qquad
X(\omega_2)=1,\qquad
X(\omega_3)=0,\qquad
X(\omega_4)=1.
$$

Thus $X=1$ means the transaction is fraudulent, and $X=0$ means it is legitimate.

Now suppose the bank does not observe the full outcome $\omega$, but only the customer's risk class. Define $Y:\Omega\to\mathbb R$ by

$$
Y(\omega_1)=0,\qquad
Y(\omega_2)=0,\qquad
Y(\omega_3)=1,\qquad
Y(\omega_4)=1.
$$

Thus:

* $Y=0$ means “low-risk customer,”
* $Y=1$ means “high-risk customer.”

Answer the following.

4. Compute $E[X\mid Y]$ explicitly and show that

5. Verify the defining property of conditional expectation by checking that for every set $A\in\sigma(Y)$,

1.Probability Space
$(\Omega, \mathfrak{F}, P)$

A probability space consists of three components: the sample space $\Omega$, the $\sigma$-algebra $\mathfrak{F}$, and the probability measure $P$.

Sample Space ($\Omega$): Given as $\Omega = \{\omega_1, \omega_2, \omega_3, \omega_4\}$.

$\sigma$-algebra ($\mathfrak{F}$): Since $\mathfrak{F} = \mathfrak{P}(\Omega)$ (the power set), it contains all possible subsets of $\Omega$. Since $|\Omega| = 4$, $\mathfrak{F}$ contains $2^4 = 16$ elements.

Probability Measure ($P$): Defined by assigning probabilities to the singletons:$P(\{\omega_1\}) = 0.50$$P(\{\omega_2\}) = 0.10$$P(\{\omega_3\}) = 0.20$$P(\{\omega_4\}) = 0.20$

$$\mathfrak{F} = \big\{ \emptyset, \{\omega_1\}, \{\omega_2\}, \{\omega_3\}, \{\omega_4\}, \{\omega_1, \omega_2\}, \{\omega_1, \omega_3\}, \{\omega_1, \omega_4\}, \{\omega_2, \omega_3\}, \{\omega_2, \omega_4\}, \{\omega_3, \omega_4\}, \{\omega_1, \omega_2, \omega_3\}, \{\omega_1, \omega_2, \omega_4\}, \{\omega_1, \omega_3, \omega_4\}, \{\omega_2, \omega_3, \omega_4\}, \Omega \big\}$$

2.Compute the $\sigma$-algebra generated by $Y$, $\sigma(Y)$

The random variable $Y$ maps the sample space to $\mathbb{R}$ based on the customer's risk class:$Y(\omega_1) = 0, \quad Y(\omega_2) = 0 \implies Y^{-1}(\{0\}) = \{\omega_1, \omega_2\}$$Y(\omega_3) = 1, \quad Y(\omega_4) = 1 \implies Y^{-1}(\{1\}) = \{\omega_3, \omega_4\}$The $\sigma$-algebra generated by $Y$ is the collection of all pre-images of Borel sets under $Y$, which corresponds to all possible unions of these fundamental blocks along with the empty set and the entire space:$$\sigma(Y) = \big\{ \emptyset, \{\omega_1, \omega_2\}, \{\omega_3, \omega_4\}, \Omega \big\}$$

3.Measurability of $E[X \mid Y]$By definition, the conditional expectation $E[X \mid Y]$ is a random variable that is $\sigma(Y)$-measurable.For a random variable to be measurable with respect to a specific $\sigma$-algebra, its value must be completely determined by the information contained within that $\sigma$-algebra. Because the finest non-empty atomic elements (or "information blocks") available in $\sigma(Y)$ are $\{\omega_1, \omega_2\}$ and $\{\omega_3, \omega_4\}$, the function $E[X \mid Y]$ cannot distinguish between individual outcomes inside those blocks.Therefore, it must take a constant value across all outcomes within $\{\omega_1, \omega_2\}$ and another constant value across all outcomes within $\{\omega_3, \omega_4\}$.

4.Compute $E[X \mid Y]$

ExplicitlySince $E[X \mid Y]$ is constant on the sets $\{\omega_1, \omega_2\}$ (where $Y=0$) and $\{\omega_3, \omega_4\}$ (where $Y=1$), we calculate the conditional expectation for each case using standard conditional probabilities:For $Y = 0$ (the set $\{\omega_1, \omega_2\}$):$$P(Y = 0) = P(\{\omega_1\}) + P(\{\omega_2\}) = 0.50 + 0.10 = 0.60$$$$E[X \mid Y = 0] = \frac{X(\omega_1)P(\{\omega_1\}) + X(\omega_2)P(\{\omega_2\})}{P(Y = 0)} = \frac{0 \cdot 0.50 + 1 \cdot 0.10}{0.60} = \frac{0.10}{0.60} = \frac{1}{6}$$For $Y = 1$ (the set $\{\omega_3, \omega_4\}$):$$P(Y = 1) = P(\{\omega_3\}) + P(\{\omega_4\}) = 0.20 + 0.20 = 0.40$$$$E[X \mid Y = 1] = \frac{X(\omega_3)P(\{\omega_3\}) + X(\omega_4)P(\{\omega_4\})}{P(Y = 1)} = \frac{0 \cdot 0.20 + 1 \cdot 0.20}{0.40} = \frac{0.20}{0.40} = \frac{1}{2}$$Thus, writing it as an indicator function form:$$E[X \mid Y] = \frac{1}{6} \mathbf{1}_{\{Y=0\}} + \frac{1}{2} \mathbf{1}_{\{Y=1\}}$$

```python
# Verification of calculations using Python
p_w1, p_w2, p_w3, p_w4 = 0.50, 0.10, 0.20, 0.20

# Case Y = 0
p_Y0 = p_w1 + p_w2
E_X_Y0 = (0 * p_w1 + 1 * p_w2) / p_Y0
print(p_Y0, E_X_Y0)

# Case Y = 1
p_Y1 = p_w3 + p_w4
E_X_Y1 = (0 * p_w3 + 1 * p_w4) / p_Y1
print(p_Y1, E_X_Y1)
```

5. Verify the Defining Property of Conditional Expectation

We need to show that $\int_A E[X \mid Y] \, dP = \int_A X \, dP$ for all $A \in \sigma(Y)$. We check this for all four sets in $\sigma(Y)$:For $A = \emptyset$: Both sides are trivially $0$.For $A = \{\omega_1, \omega_2\}$:$$\int_A E[X \mid Y] \, dP = E[X \mid Y=0] \cdot P(\{\omega_1, \omega_2\}) = \frac{1}{6} \cdot 0.60 = 0.10$$$$\int_A X \, dP = X(\omega_1)P(\{\omega_1\}) + X(\omega_2)P(\{\omega_2\}) = 0 \cdot 0.50 + 1 \cdot 0.10 = 0.10 \quad \checkmark$$For $A = \{\omega_3, \omega_4\}$:$$\int_A E[X \mid Y] \, dP = E[X \mid Y=1] \cdot P(\{\omega_3, \omega_4\}) = \frac{1}{2} \cdot 0.40 = 0.20$$$$\int_A X \, dP = X(\omega_3)P(\{\omega_3\}) + X(\omega_4)P(\{\omega_4\}) = 0 \cdot 0.20 + 1 \cdot 0.20 = 0.20 \quad \checkmark$$For $A = \Omega$:$$\int_\Omega E[X \mid Y] \, dP = 0.10 + 0.20 = 0.30$$$$\int_\Omega X \, dP = 0.10 + 0.20 = 0.30 \quad \checkmark$$

6. Find the Conditional PMF $p_{X \mid Y}(x \mid y)$

The conditional PMF is defined as $p_{X \mid Y}(x \mid y) = P(X = x \mid Y = y) = \frac{P(X=x, Y=y)}{P(Y=y)}$.For $y = 0$:$p_{X|Y}(0 \mid 0) = P(X=0 \mid Y=0) = \frac{P(\{\omega_1\})}{P(\{\omega_1, \omega_2\})} = \frac{0.50}{0.60} = \frac{5}{6}$$p_{X|Y}(1 \mid 0) = P(X=1 \mid Y=0) = \frac{P(\{\omega_2\})}{P(\{\omega_1, \omega_2\})} = \frac{0.10}{0.60} = \frac{1}{6}$For $y = 1$:$p_{X|Y}(0 \mid 1) = P(X=0 \mid Y=1) = \frac{P(\{\omega_3\})}{P(\{\omega_3, \omega_4\})} = \frac{0.20}{0.40} = \frac{1}{2}$$p_{X|Y}(1 \mid 1) = P(X=1 \mid Y=1) = \frac{P(\{\omega_4\})}{P(\{\omega_3, \omega_4\})} = \frac{0.20}{0.40} = \frac{1}{2}$Summary Table:pX∣Y​(x∣y)x=0x=1$y = 0$$\frac{5}{6}$$\frac{1}{6}$$y = 1$$\frac{1}{2}$$\frac{1}{2}$

7. Verify the Identity $E[X \mid Y=y] = \sum_{x \in \{0,1\}} x \cdot p_{X \mid Y}(x \mid y)$For $y = 0$:$$\sum_{x \in \{0,1\}} x \cdot p_{X \mid Y}(x \mid 0) = 0 \cdot \left(\frac{5}{6}\right) + 1 \cdot \left(\frac{1}{6}\right) = \frac{1}{6} = E[X \mid Y=0] \quad \checkmark$$For $y = 1$:$$\sum_{x \in \{0,1\}} x \cdot p_{X \mid Y}(x \mid 1) = 0 \cdot \left(\frac{1}{2}\right) + 1 \cdot \left(\frac{1}{2}\right) = \frac{1}{2} = E[X \mid Y=1] \quad \checkmark$$

8. Calculate the Mean Square Error

(MSE)When using the conditional expectation $E[X \mid Y]$ to estimate $X$, the MSE is given by $E[(X - E[X \mid Y])^2]$, which is equivalent to the expected conditional variance $E[\text{Var}(X \mid Y)]$.Let's calculate the conditional variance $\text{Var}(X \mid Y = y) = E[X^2 \mid Y=y] - (E[X \mid Y=y])^2$. Since $X \in \{0,1\}$, $X^2 = X$, so $E[X^2 \mid Y=y] = E[X \mid Y=y]$.$\text{Var}(X \mid Y=0) = \frac{1}{6} - \left(\frac{1}{6}\right)^2 = \frac{1}{6} - \frac{1}{36} = \frac{5}{36}$$\text{Var}(X \mid Y=1) = \frac{1}{2} - \left(\frac{1}{2}\right)^2 = \frac{1}{2} - \frac{1}{4} = \frac{1}{4} = \frac{9}{36}$Now we take the expectation over $Y$:$$\text{MSE} = E[\text{Var}(X \mid Y)] = \text{Var}(X \mid Y=0)P(Y=0) + \text{Var}(X \mid Y=1)P(Y=1)$$$$\text{MSE} = \left(\frac{5}{36}\right)(0.60) + \left(\frac{1}{4}\right)(0.40) = \frac{5}{36} \cdot \frac{3}{5} + \frac{1}{4} \cdot \frac{2}{5} = \frac{1}{12} + \frac{1}{10} = \frac{11}{60} \approx 0.1833$$




```python
# Computing MSE
var_X_Y0 = (1/6) - (1/6)**2
var_X_Y1 = (1/2) - (1/2)**2

# Calculate MSE directly using the correctly named variables
mse_with_Y = (var_X_Y0 * 0.60) + (var_X_Y1 * 0.40)
print(f"MSE with risk class information: {mse_with_Y:.4f} ({11/60})")
```

9. Constant Estimate Comparison & Law of Total VarianceIf the bank ignores $Y$ and uses a constant value (the global mean $E[X] = 0.30$), the error becomes the global variance $\text{Var}(X)$.$$E[X^2] = E[X] = 0.30$$$$\text{Var}(X) = E[X^2] - (E[X])^2 = 0.30 - (0.30)^2 = 0.30 - 0.09 = 0.21$$Relationship to the Law of Total Variance:The Law of Total Variance states:$$\text{Var}(X) = E[\text{Var}(X \mid Y)] + \text{Var}(E[X \mid Y])$$Where:$E[\text{Var}(X \mid Y)]$ is the unexplained variance (the MSE we found in Question 8 $= 0.1833$).$\text{Var}(E[X \mid Y])$ is the variance explained by knowing $Y$.Let's verify this components split:$$E[E[X \mid Y]] = \left(\frac{1}{6}\right)(0.60) + \left(\frac{1}{2}\right)(0.40) = 0.10 + 0.20 = 0.30$$$$E[(E[X \mid Y])^2] = \left(\frac{1}{6}\right)^2(0.60) + \left(\frac{1}{2}\right)^2(0.40) = \frac{1}{36} \cdot \frac{3}{5} + \frac{1}{4} \cdot \frac{2}{5} = \frac{1}{60} + \frac{6}{60} = \frac{7}{60} \approx 0.1167$$$$\text{Var}(E[X \mid Y]) = \frac{7}{60} - (0.30)^2 = \frac{7}{60} - \frac{9}{100} = \frac{35 - 27}{300} = \frac{8}{300} = \frac{2}{75} \approx 0.0267$$Checking the sum:$$E[\text{Var}(X \mid Y)] + \text{Var}(E[X \mid Y]) = \frac{11}{60} + \frac{8}{300} = \frac{55 + 8}{300} = \frac{63}{300} = 0.21 = \text{Var}(X)$$Ignoring the customer's risk class increases the estimation error from 0.1833 to 0.21.

```python
# Verification of Total Variance
var_X = 0.30 - 0.30**2
E_EX_sq = ((1/6)**2 * 0.60) + ((1/2)**2 * 0.40)
var_EX = E_EX_sq - 0.30**2

print(f"Total Variance Var(X): {var_X:.4f}")
print(f"Unexplained Variance (MSE): {mse_with_Y:.4f}")
print(f"Explained Variance Var(E[X|Y]): {var_EX:.4f}")
print(f"Sum check: {mse_with_Y + var_EX:.4f}")
```

10. Information Theory Metricsa) Entropy $H(X)$$$H(X) = - \sum_{x \in \{0,1\}} p(x) \log_2 p(x) = - \big(0.70 \log_2 0.70 + 0.30 \log_2 0.30 \big)$$b) Conditional Entropy $H(X \mid Y)$$$H(X \mid Y) = \sum_{y \in \{0,1\}} p(y) H(X \mid Y = y)$$$H(X \mid Y=0) = - \left( \frac{5}{6}\log_2\frac{5}{6} + \frac{1}{6}\log_2\frac{1}{6} \right)$$H(X \mid Y=1) = - \left( \frac{1}{2}\log_2\frac{1}{2} + \frac{1}{2}\log_2\frac{1}{2} \right) = 1\text{ bit}$$$H(X \mid Y) = 0.60 \cdot H(X \mid Y=0) + 0.40 \cdot 1$$c) Mutual Information $I(X; Y)$$$I(X; Y) = H(X) - H(X \mid Y)$$d) Percentage of Uncertainty Removed$$\text{Percentage} = \frac{I(X; Y)}{H(X)} \times 100\%$$

```python
import numpy as np

# Entropy calculation helper
def entropy(probs):
    return -sum(p * np.log2(p) for p in probs if p > 0)

# a) H(X)
H_X = entropy([0.70, 0.30])

# b) H(X|Y)
H_X_Y0 = entropy([5/6, 1/6])
H_X_Y1 = entropy([0.5, 0.5])
H_X_Y = (0.60 * H_X_Y0) + (0.40 * H_X_Y1)

# c) I(X; Y)
I_XY = H_X - H_X_Y

# d) Percentage
pct_removed = (I_XY / H_X) * 100

print(f"a) H(X)      = {H_X:.4f} bits")
print(f"b) H(X|Y)    = {H_X_Y:.4f} bits")
print(f"c) I(X; Y)   = {I_XY:.4f} bits")
print(f"d) Uncertainty Removed = {pct_removed:.2f}%")
```

# Q2) Conditional Density of the Multivariate Gaussian

Let $\mathbf{Z}$ be the concatenated random matrix
$\mathbf{Z} = \begin{bmatrix} \mathbf{X} \\ \mathbf{Y} \end{bmatrix} \in \mathbb{R}^{n+m}$.


The random variable $\mathbf{Z}$ is said to have a **joint multivariate normal distribution** if every linear combination of its components follows a univariate normal distribution.

For $\mathbf{Z}$ to possess a **joint density function** (meaning its distribution is absolutely continuous with respect to the Lebesgue measure on $\mathbb{R}^{n+m}$), its covariance matrix must be non-singular (positive definite).


The joint density function $f_{\mathbf{X}, \mathbf{Y}}(\mathbf{z})$ for $\mathbf{z} \in \mathbb{R}^{n+m}$ is given by:

$$f_{\mathbf{X}, \mathbf{Y}}(\mathbf{z}) = \frac{1}{\sqrt{(2\pi)^{n+m} |\boldsymbol{\Sigma}|}} \exp\left( -\frac{1}{2} (\mathbf{z} - \boldsymbol{\mu})^T \boldsymbol{\Sigma}^{-1} (\mathbf{z} - \boldsymbol{\mu}) \right).$$

Here:

* **Mean ($\boldsymbol{\mu}$):** $\boldsymbol{\mu} = E[\mathbf{Z}] = \begin{bmatrix} \boldsymbol{\mu}_X \\ \boldsymbol{\mu}_Y \end{bmatrix} \in \mathbb{R}^{n+m}$, where $\boldsymbol{\mu}_X \in \mathbb{R}^n$ and $\boldsymbol{\mu}_Y \in \mathbb{R}^m$.
* **Covariance Matrix ($\boldsymbol{\Sigma}$):** A symmetric, positive definite matrix $\boldsymbol{\Sigma} \in \mathbb{R}^{(n+m) \times (n+m)}$ partitioned as:
$$\boldsymbol{\Sigma} = \begin{bmatrix} \boldsymbol{\Sigma}_{XX} & \boldsymbol{\Sigma}_{XY} \\ \boldsymbol{\Sigma}_{YX} & \boldsymbol{\Sigma}_{YY} \end{bmatrix}$$
    * $\boldsymbol{\Sigma}_{XX} = \text{Var}(\mathbf{X})$ (size $n \times n$)
    * $\boldsymbol{\Sigma}_{YY} = \text{Var}(\mathbf{Y})$ (size $m \times m$)
    * $\boldsymbol{\Sigma}_{XY} = \text{Cov}(\mathbf{X}, \mathbf{Y}) = \boldsymbol{\Sigma}_{YX}^T$ (size $n \times m$)
* **Determinant ($|\boldsymbol{\Sigma}|$):** The normalization factor ensuring the total integral is 1.

1. Show that the conditional distribution is given by $f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y})$

$$f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{1}{\sqrt{(2\pi)^n |\mathbf{S}_{XX}|}} \exp\left( -\frac{1}{2} (\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) \right),$$
where
* $\boldsymbol{\Sigma}_{X\mid Y} = \boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}$
* $\boldsymbol{\mu}^* = \boldsymbol{\mu}_x + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_y)$

To derive the conditional distribution $f_{\mathbf{X}\mid\mathbf{Y}}(\mathbf{x} \mid \mathbf{y})$, we use the fundamental definition of conditional probability density:$$f_{\mathbf{X}\mid\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{f_{\mathbf{X},\mathbf{Y}}(\mathbf{x}, \mathbf{y})}{f_{\mathbf{Y}}(\mathbf{y})}$$Rather than explicitly dividing the two densities and keeping track of complex normalization constants, the standard and most elegant proof relies on a linear transformation that completely uncorrelates $\mathbf{X}$ and $\mathbf{Y}$.

We want to define a new random variable $\mathbf{X}^*$ as a linear combination of $\mathbf{X}$ and $\mathbf{Y}$ such that it is completely uncorrelated with $\mathbf{Y}$. Let's guess a form:$$\mathbf{X}^* = \mathbf{X} - \mathbf{M}\mathbf{Y}$$For $\mathbf{X}^*$ and $\mathbf{Y}$ to be uncorrelated, their cross-covariance must equal zero matrix:$$\text{Cov}(\mathbf{X}^*, \mathbf{Y}) = \mathbf{0}$$Expanding this using the linearity of covariance:$$\text{Cov}(\mathbf{X} - \mathbf{M}\mathbf{Y}, \mathbf{Y}) = \text{Cov}(\mathbf{X}, \mathbf{Y}) - \mathbf{M}\text{Var}(\mathbf{Y}) = \mathbf{0}$$$$\boldsymbol{\Sigma}_{XY} - \mathbf{M}\boldsymbol{\Sigma}_{YY} = \mathbf{0} \implies \mathbf{M} = \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}$$Thus, our transformed variable is:$$\mathbf{X}^* = \mathbf{X} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\mathbf{Y}$$

Compute Mean and Covariance of the Joint TransformationLet's represent this transformation as a joint matrix operation on $\mathbf{Z} = \begin{bmatrix} \mathbf{X} \\ \mathbf{Y} \end{bmatrix}$:$$\begin{bmatrix} \mathbf{X}^* \\ \mathbf{Y} \end{bmatrix} = \begin{bmatrix} \mathbf{I}_n & -\boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1} \\ \mathbf{0} & \mathbf{I}_m \end{bmatrix} \begin{bmatrix} \mathbf{X} \\ \mathbf{Y} \end{bmatrix} = \mathbf{A}\mathbf{Z}$$Since a linear transformation of a multivariate normal vector yields another multivariate normal vector, $\begin{bmatrix} \mathbf{X}^* \\ \mathbf{Y} \end{bmatrix}$ is jointly Gaussian.1. Mean Vector:$$E\begin{bmatrix} \mathbf{X}^* \\ \mathbf{Y} \end{bmatrix} = \begin{bmatrix} E[\mathbf{X}] - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}E[\mathbf{Y}] \\ E[\mathbf{Y}] \end{bmatrix} = \begin{bmatrix} \boldsymbol{\mu}_x - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\mu}_y \\ \boldsymbol{\mu}_y \end{bmatrix}$$2. Covariance Matrix:Using the property $\boldsymbol{\Sigma}_{\text{new}} = \mathbf{A}\boldsymbol{\Sigma}\mathbf{A}^T$:$$\text{Var}\begin{bmatrix} \mathbf{X}^* \\ \mathbf{Y} \end{bmatrix} = \begin{bmatrix} \text{Var}(\mathbf{X}^*) & \text{Cov}(\mathbf{X}^*, \mathbf{Y}) \\ \text{Cov}(\mathbf{Y}, \mathbf{X}^*) & \text{Var}(\mathbf{Y}) \end{bmatrix}$$We already designed $\mathbf{X}^*$ so that $\text{Cov}(\mathbf{X}^*, \mathbf{Y}) = \mathbf{0}$. Now let's calculate $\text{Var}(\mathbf{X}^*)$:$$\text{Var}(\mathbf{X}^*) = \text{Cov}(\mathbf{X} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\mathbf{Y}, \, \mathbf{X} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\mathbf{Y})$$$$\text{Var}(\mathbf{X}^*) = \text{Var}(\mathbf{X}) - \text{Cov}(\mathbf{X}, \mathbf{Y})\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\text{Cov}(\mathbf{Y}, \mathbf{X}) + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\text{Var}(\mathbf{Y})\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}$$$$\text{Var}(\mathbf{X}^*) = \boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX} + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}$$$$\text{Var}(\mathbf{X}^*) = \boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}$$Notice that this is precisely the definition of the Schur complement matrix given in your image: $\boldsymbol{\Sigma}_{X\mid Y}$.Therefore, the joint covariance of our transformed variables is perfectly block-diagonal:$$\text{Var}\begin{bmatrix} \mathbf{X}^* \\ \mathbf{Y} \end{bmatrix} = \begin{bmatrix} \boldsymbol{\Sigma}_{X\mid Y} & \mathbf{0} \\ \mathbf{0} & \boldsymbol{\Sigma}_{YY} \end{bmatrix}$$


Utilize IndependenceBecause $\mathbf{X}^*$ and $\mathbf{Y}$ are jointly Gaussian and their cross-covariance is zero, $\mathbf{X}^*$ and $\mathbf{Y}$ are statistically independent.Independence means that conditioning $\mathbf{X}^*$ on $\mathbf{Y}$ changes nothing about its distribution:$$f_{\mathbf{X}^*\mid\mathbf{Y}}(\mathbf{x}^* \mid \mathbf{y}) = f_{\mathbf{X}^*}(\mathbf{x}^*)$$Since $\mathbf{X}^*$ is a univariate/multivariate Gaussian variable by itself, its marginal distribution is fully determined by its mean and variance:$$\mathbf{X}^* \sim \mathcal{N}\big(\boldsymbol{\mu}_x - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\mu}_y, \, \boldsymbol{\Sigma}_{X\mid Y}\big)$$


Substitute Back to Find the Conditional DistributionNow we map this relationship back to our target variable $\mathbf{X}$. Given that $\mathbf{X} = \mathbf{X}^* + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\mathbf{Y}$, if we condition on the event that $\mathbf{Y} = \mathbf{y}$ (treating $\mathbf{y}$ as a fixed constant vector):$$\mathbf{X} \mid (\mathbf{Y}=\mathbf{y}) = \mathbf{X}^* + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\mathbf{y}$$Taking the conditional expectation and conditional variance:1. Conditional Mean ($\boldsymbol{\mu}^*$):$$\boldsymbol{\mu}^* = E[\mathbf{X} \mid \mathbf{Y}=\mathbf{y}] = E[\mathbf{X}^*] + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\mathbf{y}$$$$\boldsymbol{\mu}^* = \big(\boldsymbol{\mu}_x - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\mu}_y\big) + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\mathbf{y}$$$$\boldsymbol{\mu}^* = \boldsymbol{\mu}_x + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_y)$$2. Conditional Expectation (The Best Estimate)Using the property for joint normals:$$E[X \mid Y = y] = \mu_X + \frac{\text{Cov}(X,Y)}{\text{Var}(Y)}(y - \mu_Y)$$Plugging in our derived parameters ($\mu_X = 0, \mu_Y = 0, \text{Cov}(X,Y) = 1, \text{Var}(Y) = 5$):$$E[X \mid Y = y] = 0 + \frac{1}{5}(y - 0) = \frac{1}{5}y$$Signal-to-Total-Variance Ratio:$$\text{Ratio} = \frac{\text{Var}(\text{Signal})}{\text{Var}(\text{Total})} = \frac{\text{Var}(X)}{\text{Var}(X) + \text{Var}(Q)} = \frac{1}{1 + 4} = \frac{1}{5}$$Thus, our best estimate updates $X$ by scaling our measurement $y$ down by the signal-to-total-variance ratio.

Information Theory: Differential EntropyThe differential entropy of a continuous Gaussian random variable $W \sim \mathcal{N}(\mu, \sigma^2)$ is given by $h(W) = \frac{1}{2}\log_2(2\pi e \sigma^2)$.Compute $h(X)$:$$h(X) = \frac{1}{2}\log_2(2\pi e \cdot 1) = \frac{1}{2}\log_2(2\pi e)$$Compute $h(X \mid Y)$:First, find the conditional variance $\sigma^2_{X \mid Y}$:$$\sigma^2_{X \mid Y} = \sigma^2_X - \frac{\text{Cov}(X,Y)^2}{\text{Var}(Y)} = 1 - \frac{1^2}{5} = 1 - 0.20 = 0.80$$$$h(X \mid Y) = \frac{1}{2}\log_2(2\pi e \cdot 0.80)$$

```python
# ==========================================
# COLAB CELL: QUESTIONS 5 & 6
# ==========================================
import numpy as np

def differential_entropy(variance):
    return 0.5 * np.log2(2 * np.pi * np.e * variance)

# Baseline metrics (Var(Q) = 4)
h_X = differential_entropy(1)
var_X_Y_base = 1 - (1**2 / 5)  # 0.80
h_X_Y_base = differential_entropy(var_X_Y_base)
I_XY_base = h_X - h_X_Y_base
pct_removed_base = (I_XY_base / h_X) * 100

print("=== Base Case: Var(Q) = 4 ===")
print(f"h(X)                  = {h_X:.4f} bits")
print(f"h(X | Y)              = {h_X_Y_base:.4f} bits")
print(f"Mutual Information I  = {I_XY_base:.4f} bits")
print(f"Uncertainty Removed   = {pct_removed_base:.2f}%\n")

# High Noise Case (Var(Q) = 100)
# Total Var(Y) becomes 1 + 100 = 101. Cov(X,Y) remains 1.
var_Y_high = 101
var_X_Y_high = 1 - (1**2 / var_Y_high)
h_X_Y_high = differential_entropy(var_X_Y_high)
I_XY_high = h_X - h_X_Y_high
pct_removed_high = (I_XY_high / h_X) * 100

print("=== High Noise Case: Var(Q) = 100 ===")
print(f"h(X | Y)              = {h_X_Y_high:.4f} bits")
print(f"Mutual Information I  = {I_XY_high:.4f} bits")
print(f"Uncertainty Removed   = {pct_removed_high:.2f}%")
```
